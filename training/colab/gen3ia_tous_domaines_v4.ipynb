# 🧠 GEN3IA — Entraînement TOUS DOMAINES v4 (curriculum 1 000 000 lignes, 12 phases)

**Objectif v4 :** couvrir les **8 catégories / 12 phases** de la feuille de
route (langage, raisonnement, programmation, outils/RAG, vision, audio, vidéo,
orchestration multimodale, entraînement de modèles, sécurité défensive,
métiers professionnels, sciences) avec **1 000 000 d'exemples exactement** —
37 datasets bilingues FR/EN — PLUS des **datasets Hugging Face réels** pour
les médias (COCO train en streaming pour le photoréalisme).

| Profil GPU | Phase A (texte, 33 datasets) | Phase B (média) | Durée estimée |
|---|---|---|---|
| Tesla T4 | 3 000 pas QLoRA (cap 400k exemples) | 3 000 pas @512 | ~6 h |
| L4 | 6 000 pas QLoRA | 3 000 pas @512→640 | ~5 h |
| A100 | 12 000 pas QLoRA | 3 000 pas @512→768 | ~3 h |

**Honnêteté (inchangée) :** la Phase A entraîne le COMPORTEMENT par domaine
(format, structure, réflexes de sécurité et d'incertitude, calculs exacts) —
la connaissance encyclopédique reste celle du modèle de base. La Phase B
stylise le photoréalisme sur de vraies photos — gen3 ne devient ni Sora ni
un modèle vidéo de bout en bout. Le tout reste compatible WebGPU navigateur.

**Reprise :** checkpoints + MAX_HOURS + `--resume` — tout run interrompu
repart du dernier checkpoint vert.

```python
# ═══════════════ CELLULE A1 — PHASE A (langagière) : curriculum TOUS DOMAINES ═══════════════
# 33 datasets textuels (7 v3 + 26 v4 ≈ 946 000 exemples, 12 phases) apprennent
# à gen3 le COMPORTEMENT par phase : format, calcul exact, honnêteté, sécurité.
# Cap SAMPLE_TEXT (400 000) : la variété prime sur la répétition ; le notebook
# entraîne sur un échantillon stratifié déterministe (seed 42).
from torch.utils.data import Dataset
from transformers import AutoTokenizer

TEXT_DATASETS = ["abbreviations", "code", "knowledge", "security", "writing", "business", "multimodal",
                 "nlp_comprehension", "text_generation", "contextual_understanding",
                 "conversational_memory", "intent_recognition", "reasoning", "math",
                 "problem_solving", "code_v4", "tools", "rag_documents", "databases",
                 "image_analysis", "image_generation", "voice_speech", "audio_understanding",
                 "music_v4", "video_v4", "multimodal_v4", "ml_data", "ml_training",
                 "ml_evaluation", "ml_management", "security_v4", "professions", "sciences"]
text_rows = []
for _name in TEXT_DATASETS:
    _rows = DATASETS.get(_name, [])
    text_rows.extend(_rows)
    print(f"{_name:26s} {len(_rows):>9,}")
print("TOTAL texte brut :", f"{len(text_rows):,}")

SAMPLE_TEXT = int(os.environ.get("SAMPLE_TEXT", "400000"))
if len(text_rows) > SAMPLE_TEXT:
    import random as _rs
    _rs.Random(42).shuffle(text_rows)
    text_rows = text_rows[:SAMPLE_TEXT]
    print(f"Cap appliqué : {SAMPLE_TEXT:,} exemples stratifiés (seed 42)")

TEXT_MODEL_ID = os.environ.get("TEXT_MODEL_ID", "google/gemma-3-4b-it")
# gemma-3 est un modèle GATED : accepter la licence sur huggingface.co/google/gemma-3-4b-it
# Alternative libre : TEXT_MODEL_ID=Qwen/Qwen2.5-3B-Instruct
print("Modèle langagier :", TEXT_MODEL_ID)

tok_text = AutoTokenizer.from_pretrained(TEXT_MODEL_ID)
if tok_text.pad_token is None:
    tok_text.pad_token = tok_text.eos_token

CTX = int(os.environ.get("CTX", "1024"))

def formater_exemple(row):
    question = row["instruction"].strip()
    entree = (row.get("input") or "").strip()
    if entree:
        question = f"{question}\n\n{entree}"
    return question, row["output"].strip()

class CurriculumDataset(Dataset):
    def __init__(self, rows, tokenizer, ctx=CTX):
        self.exemples = rows
        self.tok = tokenizer
        self.ctx = ctx
    def __len__(self):
        return len(self.exemples)
    def __getitem__(self, i):
        row = self.exemples[i]
        question, cible = formater_exemple(row)
        ids_q = self.tok(question, add_special_tokens=False)["input_ids"]
        ids_r = self.tok(cible + "<end_of_turn>", add_special_tokens=False)["input_ids"]
        ids_q = [self.tok.bos_token_id] + ids_q + [self.tok.eos_token_id]
        input_ids = (ids_q + ids_r)[: self.ctx]
        labels = ([-100] * len(ids_q) + ids_r)[: self.ctx]   # loss UNIQUEMENT sur la réponse
        pad = self.ctx - len(input_ids)
        input_ids = input_ids + [self.tok.pad_token_id] * pad
        labels = labels + [-100] * pad
        return {"input_ids": torch.tensor(input_ids),
                "labels": torch.tensor(labels),
                "attention_mask": torch.tensor([1] * (self.ctx - pad) + [0] * pad)}

# budgets v4 : T4 → 3 000 | L4 → 6 000 | A100 → 12 000 pas
if "T4" in GPU_NAME or VRAM_Go < 17:
    STEPS_TEXT, BATCH_TEXT, ACCUM_TEXT = int(os.environ.get("STEPS_TEXT", "3000")), 1, 8
elif "L4" in GPU_NAME or VRAM_Go < 48:
    STEPS_TEXT, BATCH_TEXT, ACCUM_TEXT = int(os.environ.get("STEPS_TEXT", "6000")), 2, 8
else:
    STEPS_TEXT, BATCH_TEXT, ACCUM_TEXT = int(os.environ.get("STEPS_TEXT", "12000")), 4, 8
MAX_HOURS_TEXT = float(os.environ.get("MAX_HOURS", "0"))

ds_text = CurriculumDataset(text_rows, tok_text)
print(f"Dataset langagier prêt : {len(ds_text):,} exemples — {STEPS_TEXT:,} pas (batch effectif {BATCH_TEXT * ACCUM_TEXT})")
```

```python
# ═══════════════ CELLULE A2 — PHASE A : entraînement QLoRA (Gemma 3) ═══════════════
# QLoRA 4-bit : le modèle gelé en NF4, seuls r16 adapters sont entraînés (~1 %).
# Cosine schedule + warmup, comme la phase média. Reprise MAX_HOURS sécurisée.
from transformers import AutoModelForCausalLM, get_cosine_schedule_with_warmup
from peft import LoraConfig, get_peft_model, prepare_model_for_kbit_training

import torch.utils.data as _tud

TEXT_LR = float(os.environ.get("TEXT_LR", "2e-4"))
try:
    from transformers import BitsAndBytesConfig
    bnb = BitsAndBytesConfig(load_in_4bit=True, bnb_4bit_quant_type="nf4",
                             bnb_4bit_compute_dtype=torch.bfloat16)
    model_text = AutoModelForCausalLM.from_pretrained(
        TEXT_MODEL_ID, quantization_config=bnb, device_map={"": 0},
        torch_dtype=torch.bfloat16, trust_remote_code=True)
except Exception as e:
    print("⚠️ 4-bit indisponible :", e, "— repli bf16 (VRAM ≥ 16 Go nécessaire)")
    model_text = AutoModelForCausalLM.from_pretrained(
        TEXT_MODEL_ID, device_map={"": 0}, torch_dtype=torch.bfloat16,
        trust_remote_code=True)

model_text.config.use_cache = False
model_text = prepare_model_for_kbit_training(model_text)
lora_text = LoraConfig(r=16, lora_alpha=32, lora_dropout=0.05, bias="none",
                       task_type="CAUSAL_LM",
                       target_modules=["q_proj", "k_proj", "v_proj", "o_proj"])
model_text = get_peft_model(model_text, lora_text)
model_text.enable_input_require_grads()
model_text.gradient_checkpointing_enable()
model_text.print_trainable_parameters()

loader = _tud.DataLoader(ds_text, batch_size=BATCH_TEXT, shuffle=True, drop_last=True)
opt_text = torch.optim.AdamW([p for p in model_text.parameters() if p.requires_grad], lr=TEXT_LR)
sched_text = get_cosine_schedule_with_warmup(opt_text, max(10, STEPS_TEXT // 20), STEPS_TEXT)

def pas_textuelle(loader, opt, sched, steps, max_hours=0.0):
    model_text.train()
    depart = time.time()
    step, cumul = 0, 0.0
    opt.zero_grad(set_to_none=True)
    while step < steps:
        for batch in loader:
            batch = {k: v.to(model_text.device) for k, v in batch.items()}
            out = model_text(**batch)
            (out.loss / ACCUM_TEXT).backward()
            cumul += out.loss.item()
            if (step + 1) % ACCUM_TEXT == 0:
                torch.nn.utils.clip_grad_norm_([p for p in model_text.parameters() if p.requires_grad], 1.0)
                opt.step(); sched.step(); opt.zero_grad(set_to_none=True)
                step += 1
                if step % 50 == 0 or step == steps:
                    print(f"text step {step}/{steps} — loss {cumul / ACCUM_TEXT:.4f} "
                          f"({(time.time() - depart) / max(step, 1):.1f} s/pas)")
                    cumul = 0.0
                if step >= steps:
                    break
                if max_hours and (time.time() - depart) > max_hours * 3600:
                    print("⏱️ MAX_HOURS_TEXT atteint — arrêt propre à", step)
                    break
        if step >= steps:
            break
    return step

step_text = pas_textuelle(loader, opt_text, sched_text, STEPS_TEXT, MAX_HOURS_TEXT)
gc.collect(); torch.cuda.empty_cache()
print("✅ Phase A terminée :", step_text, "pas QLoRA langagiers")
```

```python
# ═══════════════ CELLULE A3 — PHASE A : test qualitatif par PHASE (12 groupes) ═══════════════
# Honnêteté par construction : on NE publie PAS un adaptateur sans l'avoir testé.
# Un échantillon par phase du curriculum — sortie attendue STRUCTURÉE.
model_text.eval()
def tester_langagier(question, max_new=280):
    ids = tok_text(question, return_tensors="pt").to(model_text.device)
    with torch.no_grad():
        out = model_text.generate(**ids, max_new_tokens=max_new, do_sample=True,
                                  temperature=0.7, top_p=0.9,
                                  pad_token_id=tok_text.pad_token_id)
    return tok_text.decode(out[0][ids["input_ids"].shape[1]:], skip_special_tokens=True)

import random as _r
_r.seed(42)
PHASES = {
    "1-langage": ["nlp_comprehension", "text_generation", "intent_recognition"],
    "2-raisonnement": ["reasoning", "math", "problem_solving"],
    "3-programmation": ["code_v4", "code"],
    "4-5-outils-rag": ["tools", "rag_documents", "databases"],
    "6-7-vision": ["image_analysis", "image_generation"],
    "8-audio": ["voice_speech", "audio_understanding", "music_v4"],
    "9-video": ["video_v4"],
    "10-orchestration": ["multimodal_v4"],
    "11-modeles": ["ml_data", "ml_training", "ml_evaluation"],
    "12-securite": ["security_v4"],
    "metiers": ["professions"],
    "sciences": ["sciences"],
}
tests = {}
for phase, domaines in PHASES.items():
    dispos = [d for d in domaines if DATASETS.get(d) and any(r["language"] == "fr" for r in DATASETS[d])]
    if not dispos:
        continue
    _row = _r.choice([r for r in DATASETS[_r.choice(dispos)] if r["language"] == "fr"])
    _question, _attendu = formater_exemple(_row)
    _reponse = tester_langagier(f"<start_of_turn>user\n{_question}<end_of_turn>\n<start_of_turn>model\n")
    tests[phase] = {"demande": _question[:200], "genere": _reponse[:400]}
    print(f"\n—— {phase} ——\nDEMANDE : {_question[:160]}…\nGÉNÉRÉ : {_reponse[:300]}…")
with open("/content/tests_langagiers_v4.json", "w", encoding="utf-8") as fh:
    json.dump(tests, fh, ensure_ascii=False, indent=2)
print("\n→ tests_langagiers_v4.json sauvé (joint à la publication)")
```

```python
# ═══════════════ CELLULE 6 — Fine-tune LoRA GÉANTS (SD 2.1-base) ═══════════════
# SD 2.1-base : licence CreativeML Open RAIL, usage commercial OK, UNet 865 M
# compatible inférence WebGPU navigateur. LoRA r32 : ~1,2 % des poids entraînés.
import random, glob
import numpy as np
from torch.utils.data import Dataset, DataLoader
from PIL import Image
from diffusers import StableDiffusionPipeline, DDPMScheduler
from peft import LoraConfig, get_peft_model
import torch.nn.functional as F

MODEL_ID = "stabilityai/stable-diffusion-2-1-base"
SEED = 20261025
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)

pipe = StableDiffusionPipeline.from_pretrained(
    MODEL_ID, torch_dtype=torch.float16, safety_checker=None,
    requires_safety_checker=False)
pipe.scheduler = DDPMScheduler.from_config(pipe.scheduler.config)
try:
    pipe.unet.set_attn_processor(pipe.unet.default_attn_processor())  # sdpa natif
except Exception:
    pass
pipe = pipe.to("cuda")
pipe.set_progress_bar_config(disable=True)
unet, vae, text_encoder, tokenizer = pipe.unet, pipe.vae, pipe.text_encoder, pipe.tokenizer
unet.enable_gradient_checkpointing()

lora_cfg = LoraConfig(
    r=LORA_R, lora_alpha=LORA_ALPHA, lora_dropout=0.05, bias="none",
    target_modules=["to_q", "to_k", "to_v", "to_out.0"],
)
unet_lora = get_peft_model(unet, lora_cfg)
unet_lora.train()
unet_lora.print_trainable_parameters()

class PhotoRealDS(Dataset):
    """Photos COCO (70 %) + briefs Gen3ia v2 repris sur photos réelles (30 %)."""
    def __init__(self, pairs, prompts, size=512):
        self.pairs, self.prompts, self.size = pairs, prompts, size
    def __len__(self):
        return len(self.pairs) + len(self.prompts)
    def __getitem__(self, i):
        if i < len(self.pairs):
            p = self.pairs[i]
            img = Image.open(p["image"]).convert("RGB")
            prompt = p["prompt"]
        else:
            img = Image.open(random.choice(self.pairs)["image"]).convert("RGB")
            prompt = random.choice(self.prompts)
        img = img.resize((self.size, self.size), Image.LANCZOS)
        arr = (torch.from_numpy(np.array(img)).float() / 127.5 - 1.0).permute(2, 0, 1)
        return {"pixel": arr, "prompt": prompt}

dl = DataLoader(PhotoRealDS(coco_pairs, photoreal_prompts, RESOLUTIONS[0]),
                batch_size=BATCH, shuffle=True, num_workers=2,
                drop_last=True, persistent_workers=True)

optimizer = torch.optim.AdamW(
    [p for p in unet_lora.parameters() if p.requires_grad],
    lr=LR, weight_decay=1e-2)

def lr_lambda(step):
    if step < WARMUP:
        return step / max(1, WARMUP)
    prog = (step - WARMUP) / max(1, MAX_STEPS - WARMUP)
    return 0.5 * (1 + math.cos(math.pi * min(1.0, prog)))
scheduler = torch.optim.lr_scheduler.LambdaLR(optimizer, lr_lambda)

noise_scheduler = DDPMScheduler.from_config(pipe.scheduler.config)
ALPHAS_CUMPROD = noise_scheduler.alphas_cumprod.clamp(min=1e-8).double()

def min_snr_weights(timesteps, gamma=MIN_SNR_GAMMA):
    """Pondération min-SNR-γ (Hang et al. 2023) — stabilise le fine-tune."""
    a = ALPHAS_CUMPROD[timesteps.detach().cpu()].to(torch.float32).to(timesteps.device)
    snr = a / (1 - a)
    return torch.clamp(snr, max=gamma) / snr

RES_PALIERS = {int(MAX_STEPS * 0.60): RESOLUTIONS[1] if len(RESOLUTIONS) > 1 else None,
               int(MAX_STEPS * 0.85): RESOLUTIONS[2] if len(RESOLUTIONS) > 2 else None}
CKPT_DIR = "/content/gen3ia_checkpoint"
os.makedirs(CKPT_DIR, exist_ok=True)
STEP_HISTORY, t0, step = [], time.time(), 0

def recharger_dataloader(res):
    return DataLoader(PhotoRealDS(coco_pairs, photoreal_prompts, res),
                      batch_size=BATCH, shuffle=True, num_workers=2,
                      drop_last=True, persistent_workers=True)

dl = recharger_dataloader(RESOLUTIONS[0])
resolution_courante = RESOLUTIONS[0]
print(f"🚀 Entraînement : {MAX_STEPS} pas — res {resolution_courante}px")

while step < MAX_STEPS:
    if step in RES_PALIERS and RES_PALIERS[step] and RES_PALIERS[step] != resolution_courante:
        resolution_courante = RES_PALIERS[step]
        dl = recharger_dataloader(resolution_courante)
        print(f"📈 Montée en résolution : {resolution_courante}px (step {step})")
    if MAX_HOURS and (time.time() - t0) > MAX_HOURS * 3600:
        print(f"⏱ Budget temps atteint ({MAX_HOURS} h) — arrêt propre au step {step}")
        break
    for batch in dl:
        if step >= MAX_STEPS:
            break
        if step in RES_PALIERS and RES_PALIERS[step] and RES_PALIERS[step] != resolution_courante:
            break  # recharger le dataloader à la bonne résolution
        # CFG dropout : 10 % de prompts vides → le UNet apprend aussi l'inconditionnel
        prompts = list(batch["prompt"])
        for i in range(len(prompts)):
            if random.random() < CFG_DROPOUT:
                prompts[i] = ""
        with torch.no_grad():
            latents = vae.encode(batch["pixel"].half().cuda()).latent_dist.sample()
            latents = latents * vae.config.scaling_factor
            ids = tokenizer(prompts, padding="max_length", max_length=77,
                            truncation=True, return_tensors="pt").input_ids
            emb = text_encoder(ids.cuda(), attention_mask=None)[0].half()
        noise = torch.randn_like(latents)
        timesteps = torch.randint(0, noise_scheduler.config.num_train_timesteps,
                                  (latents.shape[0],), device="cuda").long()
        noisy = noise_scheduler.add_noise(latents, noise, timesteps)
        pred = unet_lora(noisy, timesteps, emb).sample
        mse = ((pred.float() - noise.float()) ** 2).mean(dim=(1, 2, 3))
        w = min_snr_weights(timesteps)
        loss = (mse * w).mean() / ACCUM
        loss.backward()
        if (step + 1) % ACCUM == 0:
            torch.nn.utils.clip_grad_norm_(
                [p for p in unet_lora.parameters() if p.requires_grad], 1.0)
            optimizer.step()
            scheduler.step()
            optimizer.zero_grad(set_to_none=True)
        step += 1
        if step % 100 == 0 or step == 1:
            STEP_HISTORY.append({"step": step, "loss": round(loss.item() * ACCUM, 4),
                                 "lr": round(scheduler.get_last_lr()[0], 7),
                                 "res": resolution_courante})
            print(f"  step {step}/{MAX_STEPS} — loss {loss.item() * ACCUM:.4f} — "
                  f"lr {scheduler.get_last_lr()[0]:.2e} — res {resolution_courante} — "
                  f"{time.time() - t0:.0f}s")
        if step % 1000 == 0:
            torch.save({k: v for k, v in unet_lora.state_dict().items()
                        if "lora" in k}, f"{CKPT_DIR}/lora_step{step}.pt")
            print(f"  💾 checkpoint {step} sauvegardé")

print("✅ Fine-tune terminé —", STEP_HISTORY[-1] if STEP_HISTORY else "aucun pas")
```

```python
# ═══════════════ CELLULE 8 — Export ONNX fp16 v2 (inférence WebGPU) ═══════════════
import torch as _t
import json as _j
from diffusers import UNet2DConditionModel, AutoencoderKL, CLIPTextModel

EXPORT = "/content/gen3ia_artifacts/sd-photoreal-v2"
os.makedirs(EXPORT, exist_ok=True)

class UnetWrap(_t.nn.Module):
    def __init__(self, u):
        super().__init__(); self.u = u
    def forward(self, sample, timestep, encoder_hidden_states):
        return self.u(sample, timestep, encoder_hidden_states, return_dict=False)[0]

unet_cpu = UNet2DConditionModel.from_pretrained(MODEL_ID, subfolder="unet",
                                                torch_dtype=_t.float16)
unet_cpu.load_state_dict(unet.state_dict(), strict=False)
unet_cpu = UnetWrap(unet_cpu.half()).eval()
vae_cpu = AutoencoderKL.from_pretrained(MODEL_ID, subfolder="vae",
                                        torch_dtype=_t.float16).eval()
enc_cpu = CLIPTextModel.from_pretrained(MODEL_ID, subfolder="text_encoder",
                                        torch_dtype=_t.float16).eval()
dummy_ids = _t.ones((1, 77), dtype=_t.long)
dummy_lat = _t.randn((1, 4, 64, 64), dtype=_t.float16)
dummy_ts = _t.tensor([500], dtype=_t.float16)

with _t.no_grad():
    _t.onnx.export(enc_cpu, (dummy_ids,), f"{EXPORT}/text_encoder.onnx",
        input_names=["input_ids"],
        output_names=["last_hidden_state", "pooler_output"],
        dynamic_axes={"input_ids": {0: "batch"},
                      "last_hidden_state": {0: "batch"}}, opset_version=17)
    _t.onnx.export(unet_cpu, (dummy_lat, dummy_ts,
                              _t.randn((1, 77, 1024), dtype=_t.float16)),
        f"{EXPORT}/unet.onnx",
        input_names=["sample", "timestep", "encoder_hidden_states"],
        output_names=["out_sample"],
        dynamic_axes={"sample": {0: "batch"},
                      "encoder_hidden_states": {0: "batch"},
                      "out_sample": {0: "batch"}}, opset_version=17)
    _t.onnx.export(vae_cpu.decode, (dummy_lat,), f"{EXPORT}/vae_decoder.onnx",
        input_names=["latent"], output_names=["image"],
        dynamic_axes={"latent": {0: "batch"}, "image": {0: "batch"}},
        opset_version=17)

tokenizer.save_pretrained(EXPORT)
_j.dump({"scheduler": "ddim", "steps_browser": 26, "steps_haute": 30,
         "guidance": 5.0, "grille_haute": 768,
         "note": "SD2.1-base + LoRA r32 photoréalisme Gen3ia v2 — ort-web WebGPU"},
        open(f"{EXPORT}/client_config.json", "w"), indent=2)

for f in sorted(os.listdir(EXPORT)):
    p = os.path.join(EXPORT, f)
    if os.path.isfile(p):
        print(f"  {f:28s} {os.path.getsize(p) / 1e6:8.1f} Mo")
print("✅ Export ONNX v2 terminé")
```

```python
# ═══════════════ CELLULE 10 — Vidéo v2 : keyframes ×8 + fondus ffmpeg ═══════════════
# La stratégie v2 est celle de l'essaim gen3ia : les keyframes photoréalistes
# sont réparties sur les GPU des utilisateurs (WebGPU), la fusion xfade est
# faite côté central. Ici on VALIDE le pipeline de bout en bout sur Colab.
!apt -qq install -y ffmpeg > /dev/null 2>&1
import subprocess

scenes = [random.choice(photoreal_prompts).replace("Une photo de ", "").replace("A photo of ", "")
          for _ in range(2)]
KEYFRAMES = 8   # v2 : 8 keyframes par scène (essaim : 1 keyframe = 1 part/GPU)
for si, scene in enumerate(scenes):
    imgs = pipe([scene] * KEYFRAMES, num_inference_steps=26, guidance_scale=5.0,
                negative_prompt=NEGATIVE_PROMPT, generator=torch.Generator("cuda").manual_seed(100 + si * 17)
                ).images
    for ki, im in enumerate(imgs):
        im.save(f"/content/kf_{si}_{ki}.png")
    # fondus xfade 1.6 s entre keyframes (identique au graphe client fusion.ts)
    cmd = ["ffmpeg", "-y"]
    for ki in range(KEYFRAMES):
        cmd += ["-loop", "1", "-t", "2.6", "-i", f"/content/kf_{si}_{ki}.png"]
    filtre = ""
    for ki in range(1, KEYFRAMES):
        filtre += f"[{ki - 1}{'v' if ki > 1 else ''}][{ki}]xfade=transition=fade:duration=1.0:offset={1.6 * ki}[v{ki}];" if ki > 1 else \
                  f"[0][1]xfade=transition=fade:duration=1.0:offset=1.6[v1];"
    filtre += f"[v{KEYFRAMES - 1}]fps=24,format=yuv420p[vout]"
    cmd += ["-filter_complex", filtre, "-map", "[vout]", "-c:v", "libx264",
            "-preset", "veryfast", "-crf", "26", f"/content/scene_{si}.mp4"]
    subprocess.run(cmd, check=True, capture_output=True)
    probe = subprocess.run(["ffprobe", "-v", "error", "-show_entries", "format=duration",
                            "-of", "csv=p=0", f"/content/scene_{si}.mp4"],
                           capture_output=True, text=True)
    print(f"🎬 scène {si} : MP4 {probe.stdout.strip()}s — {KEYFRAMES} keyframes fondus")
print("✅ Pipeline vidéo v2 validé")
```
